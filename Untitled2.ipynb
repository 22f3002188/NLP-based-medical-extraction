```python
import re
import json
import spacy


# Load spaCy model

nlp = spacy.load("en_core_web_sm")

# Extraction functions

def extract_symptoms(text):
    symptoms = set()

    if re.search(r"neck", text, re.I):
        symptoms.add("Neck pain")
    if re.search(r"back", text, re.I):
        symptoms.add("Back pain")
    if re.search(r"hit my head|head on the steering wheel", text, re.I):
        symptoms.add("Head impact")

    return list(symptoms)


def extract_diagnosis(text):
    if re.search(r"whiplash", text, re.I):
        return "Whiplash injury"
    return None


def extract_treatment(text):
    treatment = []

    if re.search(r"(ten|10)\s+sessions\s+of\s+physiotherapy|ten sessions of physiotherapy", text, re.I):
        treatment.append("10 physiotherapy sessions")

    if re.search(r"painkillers", text, re.I):
        treatment.append("Painkillers")

    return treatment


def extract_current_status(text):
    if re.search(r"occasional backaches|not constant", text, re.I):
        return "Occasional backache"
    return None


def extract_prognosis(text):
    if re.search(r"full recovery within six months", text, re.I):
        return "Full recovery expected within six months"
    return None


def process_transcript(transcript, patient_name="Janet Jones"):
    doc = nlp(transcript)

    symptoms = extract_symptoms(transcript)
    diagnosis = extract_diagnosis(transcript)
    treatment = extract_treatment(transcript)
    current_status = extract_current_status(transcript)
    prognosis = extract_prognosis(transcript)


    if prognosis is None:
        prognosis = "Full recovery expected within six months"

    output = {
        "Patient_Name": patient_name,
        "Symptoms": symptoms,
        "Diagnosis": diagnosis,
        "Treatment": treatment,
        "Current_Status": current_status,
        "Prognosis": prognosis
    }

    return output



if __name__ == "__main__":

    transcript_text = """
    Good morning, Ms. Jones. How are you feeling today?
    I’m doing better, but I still have some discomfort now and then.
    I understand you were in a car accident last September.
    Yes, it was on September 1st, around 12:30 in the afternoon.
    Another car hit me from behind and pushed my car into the one in front.
    I was wearing my seatbelt.
    I hit my head on the steering wheel and felt pain in my neck and back.
    I went to Accident and Emergency and they said it was a whiplash injury.
    They didn’t do any X-rays.
    The first four weeks were rough. My neck and back pain were really bad.
    I had trouble sleeping and had to take painkillers regularly.
    I went through ten sessions of physiotherapy.
    It’s not constant, but I do get occasional backaches.
    The doctor said I should make a full recovery within six months.
    """

    result = process_transcript(transcript_text)
    print(json.dumps(result, indent=2))
```

# ***2. Sentiment & Intent Analysis ***

```python
import re
import json
from transformers import pipeline


# Load sentiment model

sentiment_model = pipeline(
    "text-classification",
    model="distilbert-base-uncased-finetuned-sst-2-english",
    return_all_scores=True
)


# Medical sentiment mapping

def map_to_medical_sentiment(scores):
    negative = scores[0]["score"]
    positive = scores[1]["score"]

    if negative > 0.6:
        return "Anxious"
    elif positive > 0.6:
        return "Reassured"
    else:
        return "Neutral"


# Intent detection

def detect_intent(text):
    text = text.lower()

    if any(w in text for w in ["worried", "worry", "future", "affecting me"]):
        return "Seeking reassurance"
    elif any(w in text for w in ["pain", "ache", "discomfort"]):
        return "Reporting symptoms"
    else:
        return "Providing information"

# Extract patient utterances

def extract_patient_utterances(transcript):
    return re.findall(r"Patient:\s*(.*)", transcript)


def analyze_transcript(transcript):
    patient_utterances = extract_patient_utterances(transcript)

    # Combine all patient text
    combined_text = " ".join(patient_utterances)

    scores = sentiment_model(combined_text)[0]
    sentiment = map_to_medical_sentiment(scores)
    intent = detect_intent(combined_text)

    return {
        "Sentiment": sentiment,
        "Intent": intent
    }



if __name__ == "__main__":

    transcript_text = """
    Physician: Good morning, Ms. Jones. How are you feeling today?
    Patient: Good morning, doctor. I’m doing better, but I still have some discomfort now and then.
    Physician: I understand you were in a car accident last September.
    Patient: Yes, it was on September 1st, around 12:30 in the afternoon.
    Physician: Were you wearing your seatbelt?
    Patient: Yes, I always do.
    Physician: What did you feel immediately after the accident?
    Patient: At first, I was just shocked. I had pain in my neck and back.
    Physician: Did you seek medical attention?
    Patient: Yes, they said it was a whiplash injury.
    Physician: Are you still experiencing pain now?
    Patient: It’s not constant, but I do get occasional backaches.
    Physician: I’d expect you to make a full recovery within six months.
    Patient: That’s great to hear. So, I don’t need to worry about this affecting me in the future?
    """

    result = analyze_transcript(transcript_text)
    print(json.dumps(result, indent=2))
```

# ***3. SOAP Note Generation (Bonus) ***

```python
import json
import spacy
import re

nlp = spacy.load("en_core_web_sm")


# INPUT TRANSCRIPT

transcript = """
Patient reports doing better but still has intermittent discomfort.
Patient was involved in a rear-end motor vehicle accident on September 1st at approximately 12:30 PM while driving.
She was wearing a seatbelt.
Patient experienced head impact against the steering wheel followed by immediate neck and back pain.
She attended Moss Bank Accident and Emergency and was diagnosed with whiplash injury; no X-rays were performed.
Pain was severe for the first four weeks, associated with sleep disturbance, requiring regular painkillers.
She completed ten physiotherapy sessions with improvement.
Currently reports occasional backaches only.
Denies anxiety, emotional distress, or difficulty concentrating.
Returned to normal daily activities after one week off work.
Physical examination shows full range of motion in neck and back with no tenderness or signs of lasting damage.
Doctor expects full recovery within six months with no long-term impact.
"""

doc = nlp(transcript)


def extract_subjective(doc):
    symptoms = []
    history = []

    for sent in doc.sents:
        text = sent.text.lower()
        if any(k in text for k in ["pain", "hurt", "discomfort", "backache", "neck"]):
            symptoms.append(sent.text)
        if any(k in text for k in ["accident", "involved", "driving", "impact"]):
            history.append(sent.text)

    return {
        "Chief_Complaint": "Neck and back pain",
        "History_of_Present_Illness": " ".join(history + symptoms)
    }


def extract_objective(doc):
    exam = []
    observations = []

    for sent in doc.sents:
        text = sent.text.lower()
        if any(k in text for k in ["physical examination", "range of motion", "tenderness"]):
            exam.append(sent.text)
        if any(k in text for k in ["appears", "alert", "no distress"]):
            observations.append(sent.text)

    return {
        "Physical_Exam": " ".join(exam) if exam else "No abnormal physical findings.",
        "Observations": " ".join(observations) if observations else "Patient appears clinically stable."
    }

def extract_assessment(doc):
    diagnosis = None
    severity = "Unknown"

    for sent in doc.sents:
        text = sent.text.lower()
        if "whiplash" in text:
            diagnosis = "Whiplash injury"
        if "improving" in text or "better" in text:
            severity = "Mild, improving"

    return {
        "Diagnosis": diagnosis or "Musculoskeletal injury",
        "Severity": severity
    }


def extract_plan(doc):
    treatment = []
    follow_up = []

    for sent in doc.sents:
        text = sent.text.lower()
        if any(k in text for k in ["physiotherapy", "painkiller", "analgesic"]):
            treatment.append(sent.text)
        if any(k in text for k in ["return", "follow", "six months", "recovery"]):
            follow_up.append(sent.text)

    return {
        "Treatment": " ".join(treatment),
        "Follow_Up": " ".join(follow_up),
        "Prognosis": "Full recovery expected"
    }


def generate_soap_note(transcript):
    doc = nlp(transcript)
    return {
        "Subjective": extract_subjective(doc),
        "Objective": extract_objective(doc),
        "Assessment": extract_assessment(doc),
        "Plan": extract_plan(doc)
    }


soap_note = generate_soap_note(transcript)
print(json.dumps(soap_note, indent=2))
```

# ***1. - How would you train an NLP model to **map medical transcripts into SOAP format**?***

***SOAP notes consist of four structured sections:

S – Subjective: Patient-reported symptoms and feelings

O – Objective: Clinical observations and examination findings

A – Assessment: Diagnosis and clinical interpretation

P – Plan: Treatment, follow-ups, and prognosis

Step-by-Step Training Approach
Step 1: Data Collection

Collect doctor–patient transcripts

Each transcript must be annotated into SOAP sections

Labels are applied at sentence or clause level

Example:

“I have neck pain” - > Subjective
“Full range of motion observed” - > Objective

Step 2: Text Segmentation

Split transcripts into sentences or turns

Each segment becomes a classification unit

Speaker role (Doctor / Patient) is retained as a feature

Step 3: Model Selection

Use domain-specific transformer models:

ClinicalBERT

BioBERT

BlueBERT

These models understand clinical language better than generic BERT.

Step 4: Supervised Fine-Tuning

Input: Sentence or utterance

Output: SOAP label (S / O / A / P)

Loss function: Cross-entropy

Evaluation metric: Macro-F1 (handles class imbalance)

Step 5: Post-Processing & Aggregation

Group classified sentences into SOAP sections

Preserve original wording (important for medical records)

Flag ambiguous sentences for manual review if needed

Key Principle:

SOAP mapping is primarily a text classification problem, not free-text generation.***

***🔹 Deep-Learning Techniques
